```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid")
INPUT_PATH = "retail_store_inventory.csv"
OUTPUT_DIR = "eda_outputs"
CLEANED_CSV_PATH = "retail_store_inventory_cleaned.csv"
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
df = pd.read_csv(INPUT_PATH)
print(f"Shape: {df.shape[0]} rows x {df.shape[1]} columns")
print(df.head())
print(df.dtypes)
```

```python
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
numeric_cols = ["Inventory Level", "Units Sold", "Units Ordered", "Demand Forecast",
                 "Price", "Discount", "Competitor Pricing"]
categorical_cols = ["Store ID", "Category", "Region", "Weather Condition",
                     "Holiday/Promotion", "Seasonality"]

print(df[numeric_cols].describe().T)
for col in categorical_cols:
    print(df[col].value_counts())
```

```python
fig, axes = plt.subplots(3, 3, figsize=(16, 12))
axes = axes.flatten()
for i, col in enumerate(numeric_cols):
    sns.histplot(df[col], kde=True, ax=axes[i], color="steelblue")
    axes[i].set_title(f"Distribution: {col}")
for j in range(len(numeric_cols), len(axes)):
    fig.delaxes(axes[j])
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/numeric_distributions.png", dpi=120)
plt.close()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
axes = axes.flatten()
for i, col in enumerate(categorical_cols):
    df[col].value_counts().plot(kind="bar", ax=axes[i], color="salmon")
    axes[i].set_title(f"Counts: {col}")
    axes[i].tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/categorical_counts.png", dpi=120)
plt.close()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
axes = axes.flatten()
for i, col in enumerate(categorical_cols):
    df[col].value_counts().plot(kind="bar", ax=axes[i], color="salmon")
    axes[i].set_title(f"Counts: {col}")
    axes[i].tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/categorical_counts.png", dpi=120)
plt.close()
```

```python
for col in numeric_cols:
    Q1, Q3 = df[col].quantile(0.25), df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
    n_outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: bounds=({lower:.2f}, {upper:.2f}), outliers={n_outliers}")

fig, axes = plt.subplots(3, 3, figsize=(16, 12))
axes = axes.flatten()
for i, col in enumerate(numeric_cols):
    sns.boxplot(x=df[col], ax=axes[i], color="lightgreen")
    axes[i].set_title(f"Boxplot: {col}")
for j in range(len(numeric_cols), len(axes)):
    fig.delaxes(axes[j])
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/boxplots.png", dpi=120)
plt.close()
```

```python
corr = df[numeric_cols].corr()
print(corr.round(2))

plt.figure(figsize=(9, 7))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/correlation_heatmap.png", dpi=120)
plt.close()

sales_trend = df.groupby("Date")["Units Sold"].sum()
plt.figure(figsize=(14, 5))
sales_trend.plot(color="darkorange")
plt.title("Total Units Sold Over Time")
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/sales_trend.png", dpi=120)
plt.close()
```

```python
df_clean = df.copy()
df_clean = df_clean.drop_duplicates()
df_clean = df_clean.dropna(subset=["Date"])
```

```python
neg_mask = df_clean["Demand Forecast"] < 0
df_clean.loc[neg_mask, "Demand Forecast"] = 0
```

```python
text_cols = ["Store ID", "Product ID", "Category", "Region", "Weather Condition", "Seasonality"]
for col in text_cols:
    df_clean[col] = df_clean[col].astype(str).str.strip()
```

```python
df_clean["Holiday/Promotion"] = df_clean["Holiday/Promotion"].astype(int)
df_clean["Discount"] = df_clean["Discount"].astype(int)
for col in ["Inventory Level", "Units Sold", "Units Ordered"]:
    df_clean[col] = df_clean[col].astype(int)
```

```python
for col in ["Demand Forecast", "Price", "Competitor Pricing"]:
    Q1, Q3 = df_clean[col].quantile(0.25), df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
    df_clean[col] = df_clean[col].clip(lower, upper)
```

```python
df_clean["Year"] = df_clean["Date"].dt.year
df_clean["Month"] = df_clean["Date"].dt.month
df_clean["Day"] = df_clean["Date"].dt.day
df_clean["Weekday"] = df_clean["Date"].dt.day_name()
df_clean["Stockout_Flag"] = (df_clean["Inventory Level"] <= df_clean["Units Sold"]).astype(int)
df_clean["Sell_Through_Rate"] = (df_clean["Units Sold"] / df_clean["Inventory Level"].replace(0, np.nan)).round(3)
df_clean["Forecast_Error"] = (df_clean["Demand Forecast"] - df_clean["Units Sold"]).round(2)
df_clean["Price_Diff_vs_Competitor"] = (df_clean["Price"] - df_clean["Competitor Pricing"]).round(2)
df_clean["Net_Price"] = (df_clean["Price"] * (1 - df_clean["Discount"] / 100)).round(2)
```

```python
df_clean = df_clean.drop_duplicates().reset_index(drop=True)


df_clean.to_csv(CLEANED_CSV_PATH, index=False)
print(f"Cleaned CSV saved to: {CLEANED_CSV_PATH}")
print(df_clean.head())
```

```python
df_clean.to_csv("retail_store_inventory_cleaned.csv", index=False)
print("File saved successfully at:", os.path.abspath("retail_store_inventory_cleaned.csv"))
```
