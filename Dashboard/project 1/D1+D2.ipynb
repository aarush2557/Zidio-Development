```python
# ---------------------------------------------------------------------------
# STEP 8: Time-based train/test split
# ---------------------------------------------------------------------------
# IMPORTANT: this is time-series data, so we NEVER use a random split
# (that would let the model "see the future"). Instead we pick a cutoff date: 
# everything before it is train, everything after is test.
 
df = df.sort_values(["Store ID", "Product ID", "Date"])
 
HORIZON_DAYS = 56  # ~8 weeks - matches the forecast horizon in the brief
cutoff_date = df["Date"].max() - pd.Timedelta(days=HORIZON_DAYS)
 
train = df[df["Date"] <= cutoff_date].copy()
test = df[df["Date"] > cutoff_date].copy()
 
print(f"Cutoff date: {cutoff_date.date()}")
print(f"Train: {len(train):,} rows ({train['Date'].min().date()} to {train['Date'].max().date()})")
print(f"Test:  {len(test):,} rows ({test['Date'].min().date()} to {test['Date'].max().date()})")
 
train.to_csv("train.csv", index=False)
test.to_csv("test.csv", index=False)
 
```

```python
# ---------------------------------------------------------------------------
# STEP 9: WAPE - the accuracy metric used throughout this project
# ---------------------------------------------------------------------------
def wape(actual: pd.Series, predicted: pd.Series) -> float:
    """
    Weighted Absolute Percentage Error = total absolute error / total actual demand.
    Lower is better. Robust to low-selling products (unlike plain MAPE).
    """
    return (actual - predicted).abs().sum() / actual.sum()
 
```

```python
# ---------------------------------------------------------------------------
# STEP 10: Baseline 1 - Seasonal-naive (same store+product, 1 year earlier)
# ---------------------------------------------------------------------------
test["lookup_date"] = test["Date"] - pd.DateOffset(years=1)
full_lookup = df.set_index(["Store ID", "Product ID", "Date"])["Units Sold"]
 
def seasonal_naive_lookup(row):
    key = (row["Store ID"], row["Product ID"], row["lookup_date"])
    return full_lookup.get(key)
 
test["baseline_seasonal_naive"] = test.apply(seasonal_naive_lookup, axis=1)
 
wape_seasonal = wape(test["Units Sold"], test["baseline_seasonal_naive"])
print(f"Baseline 1 - Seasonal-naive WAPE: {wape_seasonal:.4f} ({wape_seasonal*100:.1f}% error)")
```

```python
# ---------------------------------------------------------------------------
# STEP 11: Baseline 2 - Historical average (computed ONLY on train data)
# ---------------------------------------------------------------------------
avg_lookup = train.groupby(["Store ID", "Product ID"])["Units Sold"].mean()
 
test["baseline_historical_avg"] = test.apply(
    lambda r: avg_lookup.loc[(r["Store ID"], r["Product ID"])], axis=1
)
 
wape_avg = wape(test["Units Sold"], test["baseline_historical_avg"])
print(f"Baseline 2 - Historical-average WAPE: {wape_avg:.4f} ({wape_avg*100:.1f}% error)")
 
```

```python
# ---------------------------------------------------------------------------
# STEP 12: Pick the winning baseline - "the bar" any real model must beat
# ---------------------------------------------------------------------------
winner = "Seasonal-naive" if wape_seasonal < wape_avg else "Historical-average"
print(f"\nWinning baseline: {winner}")
print("Any forecasting model built later must beat this WAPE using")
print("rolling-origin backtesting (not this single static split) to confirm it.")
 
test.to_csv("test_with_baselines.csv", index=False)
 
```
